```python
import glob
import json
import os

import numpy as np
import pandas as pd
```

```python
datasets = [
    'hmof', 'hmofheat', 'cof', 'ppn',
    'zeolite', 'c3h6c3h8coremof', 'ch4n2',
    'corecof',
]
```

```python
ds_tasks = {
    ds: [i.split('-to_')[-1] for i in glob.glob(f'../experiments/evaluate/few_shot/{ds}/*scratch*')]
    for ds in datasets
}
```

## Overview of adsorption tasks

```python
tasks_list = []
id_list = []
ds_list = []

for ds_id in ds_tasks.keys():
    for t in ds_tasks[ds_id]:
        id_lbl = f'\\texttt{{{ds_id}}}'
        ds_lbl = pretty_labels[ds_id]
        #ditto = r'\dittosymbol{}'
        
        tasks_list.append(pretty_labels[t])
        id_list.append(id_lbl)
        ds_list.append(ds_lbl)
        #id_list.append(id_lbl if id_lbl not in id_list else ditto)
        #ds_list.append(ds_lbl if ds_lbl not in ds_list else ditto)
```

```python
df = pd.DataFrame(dict(task=tasks_list, identifiers=id_list, dataset=ds_list))
```

```python
df.set_index('task', inplace=True)
print(df.to_latex())
```

## Overview of benchmark protocol

```python
tasks_list = []
epochs_list = []
layers_list = []
ds_list = []

epochs_layers = {
    'hmof': [50, 6],
    'hmofheat': [50, 7],
    'cof': [500, 6],
    'ppn': [100, 2],
    'zeolite': [500, 1],
    'c3h6c3h8coremof': [1000, 'N/A'],
    'ch4n2': [1000, 2],
    'corecof': ['N/A', 'N/A'],
}
    

for ds_id in ds_tasks.keys():
    for t in ds_tasks[ds_id]:
        n_epochs, n_frozen = epochs_layers[ds_id]

        if ds_id == 'c3h6c3h8coremof':
            n_frozen = 1 if t in [
                'C3H8_C3H6_Selectivity_1bar',
                'C3H8_C3H6_Selectivity_infinite',
                'C3H6_loadings',
            ] else 2
        
        tasks_list.append(pretty_labels[t])
        epochs_list.append(n_epochs)
        layers_list.append(n_frozen)
        ds_list.append(pretty_labels[ds_id])
```

```python
df = pd.DataFrame(dict(task=tasks_list, epochs=epochs_list, layers=layers_list, dataset=ds_list))
```

```python
df.set_index('task', inplace=True)
print(df.to_latex())
```

## Overview of few-shot protocol

```python
tasks_list = []
epochs_list = []
layers_list = []
ds_list = []

epochs_layers = {
    'hmof': [1000, 6],
    'hmofheat': [500, 7],
    'cof': [1000, 6],
    'ppn': [1000, 2],
    'zeolite': [1000, 1],
    'c3h6c3h8coremof': [1000, 2],
    'ch4n2': [1000, 2],
    'corecof': [1000, 2],
}
    

for ds_id in ds_tasks.keys():
    for t in ds_tasks[ds_id]:
        n_epochs, n_frozen = epochs_layers[ds_id]
        
        tasks_list.append(pretty_labels[t])
        epochs_list.append(n_epochs)
        layers_list.append(n_frozen)
        ds_list.append(pretty_labels[ds_id])
```

```python
df = pd.DataFrame(dict(task=tasks_list, epochs=epochs_list, layers=layers_list, dataset=ds_list))
```

```python
df.set_index('task', inplace=True)
print(df.to_latex())
```
